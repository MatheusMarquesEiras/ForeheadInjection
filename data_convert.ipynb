```python
import pandas as pd
import json
```

```python
caminho_arquivo_json = './backend/data.json'
caminho_arquivo_parquet = './backend/data.parquet'
```

```python
try:
    with open(caminho_arquivo_json, 'r', encoding='utf-8') as f:
        json_data = json.load(f)
except FileNotFoundError:
    print("Erro: O arquivo 'data.json' não foi encontrado.")
    exit()
```

```python
df_courses = pd.DataFrame(json_data['courses'])
df_topics = pd.DataFrame(json_data['topics'])
df_contents = pd.DataFrame(json_data['contents'])
```

```python
df_courses['courses'] = df_courses.apply(lambda row: row.to_dict(), axis=1)
df_topics['topics'] = df_topics.apply(lambda row: row.to_dict(), axis=1)
df_contents['contents'] = df_contents.apply(lambda row: row.to_dict(), axis=1)
```

```python
df_total = pd.concat([df_courses['course'], df_topics['topics'], df_contents['contents']], axis=1)
df_total.count()
```

```python
df_total.to_parquet(caminho_arquivo_parquet, index=False, )
```
